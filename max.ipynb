```python

even=0
odd=0
n=int(input(" enter the no of number to input"))
arr = [int(input("enter the number:"))for i in range(n)]

for num in arr:
    if num%2==1:
        odd +=1
    else:
        even +=1
arr.sort()
print(f"""Array;{arr}
sum: {sum(arr)}
avg{sum(arr)/len(arr)}
max: {max(arr)}
 min: {min(arr)}
2 nd highest: {arr[-2]}
2 nd lowest: { arr[1]}
odd numbers: {odd}
even numbers: {even}
""")
```
