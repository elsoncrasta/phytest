```python
import matplotlib.pyplot as plt
Months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
Rainfall = [25, 18, 30, 45, 60, 80, 120, 110, 70, 50, 35, 20]
plt.figure(figsize=(10, 6))
plt.plot(Months, Rainfall, marker='o', linestyle='-', color='b')
plt.title('Monthly Rainfall ')
plt.xlabel('Month')
plt.ylabel('Rainfall')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

Brands = ['Ford', 'BMW', 'Hyundai']
Sales = [250, 300, 180]

plt.figure(figsize=(8, 5))
plt.bar(Brands, Sales, color=['blue', 'green', 'red'])
plt.title('Car Sales by Brand')
plt.xlabel('Brand')
plt.ylabel('Number of Cars Sold')
plt.grid(axis='y', linestyle='--', alpha=1)
plt.show()
```

```python
import matplotlib.pyplot as plt

countries = ['USA', 'India', 'Japan', 'Germany', 'Brazil']
speeds = [70, 40, 120, 85, 55]

plt.figure(figsize=(10, 6))
plt.barh(countries, speeds, color=['purple', 'orange', 'green', 'blue', 'red'])
plt.xlabel('Average Internet Speed (Mbps)')
plt.ylabel('Country')
plt.title('Average Internet Speed by Country')
plt.grid(axis='x', linestyle='--', alpha=1)
plt.show()
```

```python
import matplotlib.pyplot as plt
labels = ['Food', 'Rent', 'Travel', 'Entertainment', 'Savings']
sizes = [5000, 8000, 2000, 1500, 2500]
colors = ['gold', 'lightcoral', 'lightskyblue', 'lightgreen', 'pink']
explode = (0, 0.1, 0, 0, 0)
plt.figure(figsize=(8, 8))
plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%',
        shadow=True, startangle=140)
plt.axis('equal')
plt.title('Monthly Expenditure Distribution')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.random.randn(100)
y = np.random.randn(100)
plt.figure(figsize=(8, 6))
plt.scatter(x, y, alpha=1, color='purple')
plt.title('Scatter Plot of 100 Random Points ')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

Heights = [150, 155, 160, 165, 170, 175, 180, 185]
Weights = [50, 55, 58, 60, 65, 72, 78, 85]

plt.figure(figsize=(8, 6))
plt.scatter(Heights, Weights, color='green', s=100, alpha=0.8, edgecolors='black')
plt.title('Heights vs. Weights of 8 People')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

Days = [1, 2, 3, 4, 5, 6, 7]
Prices = [120, 125, 123, 128, 135, 133, 140]

plt.figure(figsize=(10, 6))
plt.plot(Days, Prices, marker='o', linestyle='-', color='red', linewidth=2, markersize=8)
plt.title('Company Stock Prices Over 7 Days')
plt.xlabel('Day')
plt.ylabel('Stock Price')
plt.grid(True)
plt.xticks(Days)
plt.show()
```

```python
import matplotlib.pyplot as plt

platforms = ['Instagram', 'Facebook', 'Twitter', 'LinkedIn']
user_base_shares = [30, 40, 20, 10]
colors = ['blue', 'green', 'red', 'purple']
plt.figure(figsize=(8, 8))
plt.pie(user_base_shares, labels=platforms, colors=colors, autopct='%1.1f%%',
        shadow=True, startangle=90)
plt.axis('equal')
plt.title('Share of Social Media Platforms by User Base')
plt.show()
```

```python
import matplotlib.pyplot as plt
week_days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temperatures = [30, 32, 35, 33, 31, 29, 28]
electricity_consumption = [10, 12, 11, 14, 13, 9, 8]
fig, axes = plt.subplots(2, 1, figsize=(10, 8))
axes[0].plot(week_days, temperatures, marker='o', linestyle='-', color='orange')
axes[0].set_title('Temperature Variation Over a Week')
axes[0].set_ylabel('Temperature (°C)')
axes[0].grid(True)
axes[1].bar(week_days, electricity_consumption, color='teal')
axes[1].set_title('Daily Electricity Consumption (kWh)')
axes[1].set_xlabel('Day of the Week')
axes[1].set_ylabel('Consumption (kWh)')
axes[1].grid(axis='y', linestyle='--', alpha=1)
plt.tight_layout()
plt.show()
```
