```python
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
v=np.vstack((a, b))
print("Vertical Stack:",v)
h=np.hstack((a, b))
print("Horizontal Stack:",h)
```

```python
import numpy as np
arr=np.arange(12)
arr1=np.array_split(arr, 3)
print("Split into 3 parts:")
for i, part in enumerate(arr1):
    print(f"Part {i+1}: {part}")
```

```python
import numpy as np
arr = np.array([9, 2, 4, 7, 5])
arr1 = np.sort(arr)
print("ascending order:", arr1)
arrd=np.sort(arr)[::-1]
print(" descending order:", arrd)
```

```python
import numpy as np
arr=np.random.randint(1, 51, size=10)
great=arr[arr > 25]
print(" greater than 25:",great)
```

```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
arr1 = np.where(arr > 30, 100, arr)
print("array:",arr1)
```

```python
import numpy as np
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8], [9, 10]])
v_stack=np.vstack((arr1, arr2))
print("V Stack:", v_stack)
con=np.concatenate((arr1, arr2), axis=0)
print("Vertical", con)
```

```python
import numpy as np
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
mod = np.where(arr2 > 3, -1, arr2)
con = np.concatenate((arr1, mod))
print("Array:", con)
```
